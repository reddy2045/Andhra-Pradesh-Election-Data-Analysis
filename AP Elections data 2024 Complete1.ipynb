```python
#Top 10 Parties by Number of Candidates
plt.figure(figsize=(12,6))
df["Party"].value_counts().head(10).plot(kind="bar")
plt.title("Top 10 Parties by Number of Candidates")
plt.xlabel("Party")
plt.ylabel("Number of Candidates")
plt.xticks(rotation=90)
plt.show()
```

```python
#Party-wise Total Votes
party_votes = df.groupby("Party")["Total Votes"].sum().sort_values(ascending=False).head(10)

party_votes.plot(kind="bar", figsize=(12,6))
plt.title("Top 10 Parties by Total Votes")
plt.xlabel("Party")
plt.ylabel("Total Votes")
plt.xticks(rotation=90)
plt.show()
```

```python
#Top 10 Candidates by Votes
top10 = df.nlargest(10, "Total Votes")
plt.figure(figsize=(12,6))
sns.barplot(data=top10, x="Total Votes", y="Candidate", palette="viridis")
plt.title("Top 10 Candidates by Total Votes")
plt.tight_layout()
plt.show()
```

```python
#State wise Total Votes
plt.figure(figsize=(14,6))
df.groupby("State")["Total Votes"].sum().sort_values().plot(kind="barh", color="coral")
plt.title("State wise Total Votes")
plt.tight_layout()
plt.show()
```

```python
#EVM vs Postal Votes Comparison
plt.figure(figsize=(8,5))
plt.scatter(df["EVM Votes"], df["Postal Votes"], alpha=0.5, color="purple")
plt.title("EVM Votes vs Postal Votes")
plt.xlabel("EVM Votes")
plt.ylabel("Postal Votes")
plt.show()
```

```python
#Winner Analysis
winners = df.groupby("Assembly Constituency").apply(
    lambda x: x.loc[x["Total Votes"].idxmax()]
).reset_index(drop=True)

print(winners[["Candidate", "Party", "Total Votes", "Assembly Constituency", "State"]])
```

```python
#Export Results
# Winners Excel లో save చేయండి
winners.to_csv("election_winners.csv", index=False)
print("File saved!")
```
